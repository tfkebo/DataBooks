# Import Libraries & Load Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("data/imdb_top_1000.csv")

# Preview Dataset
display(df.head())

# Dataset Size
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

# Dataset Overview

```python
# Column Names
print(df.columns.tolist())

# Dataset Information
df.info()

# Statistical Summary
display(df.describe())
```

# Data Cleaning

```python
# Missing Values
display(df.isnull().sum())

# Duplicate Rows
print("Duplicate Rows:", df.duplicated().sum())

# Remove Duplicates
df = df.drop_duplicates()

print("Dataset Shape After Cleaning:", df.shape)
```

# Movie Ratings Analysis

```python
# IMDb Rating Statistics
display(df["IMDB_Rating"].describe())

# Rating Distribution
plt.figure(figsize=(8,5))

plt.hist(df["IMDB_Rating"], bins=15)

plt.title("Distribution of IMDb Ratings")
plt.xlabel("IMDb Rating")
plt.ylabel("Number of Movies")

plt.show()
```

# Genre Analysis

```python
# Most Common Genres
genre_counts = df["Genre"].value_counts().head(10)

display(genre_counts)

plt.figure(figsize=(10,5))

genre_counts.plot(kind="bar")

plt.title("Top 10 Movie Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Movies")

plt.show()
```

# Runtime Analysis

```python
# Runtime Statistics
display(df["Runtime"].describe())

plt.figure(figsize=(8,5))

plt.hist(df["Runtime"], bins=20)

plt.title("Distribution of Movie Runtime")
plt.xlabel("Runtime (minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

# Directors & Movie Certificates

```python
# Top 10 Directors
top_directors = df["Director"].value_counts().head(10)

display(top_directors)

plt.figure(figsize=(10,5))

top_directors.plot(kind="bar")

plt.title("Top 10 Directors by Number of Movies")
plt.xlabel("Director")
plt.ylabel("Number of Movies")

plt.show()


# Movie Certificates
certificate_counts = df["Certificate"].value_counts()

display(certificate_counts)

plt.figure(figsize=(8,5))

certificate_counts.plot(kind="bar")

plt.title("Movie Certificates")
plt.xlabel("Certificate")
plt.ylabel("Number of Movies")

plt.show()
```

# Gross Revenue Analysis

```python
# Gross Revenue Statistics
display(df["Gross"].describe())

# Remove missing values
gross_data = df["Gross"].dropna()

plt.figure(figsize=(8,5))

plt.hist(gross_data, bins=20)

plt.title("Distribution of Gross Revenue")
plt.xlabel("Gross Revenue")
plt.ylabel("Number of Movies")

plt.show()
```

# Correlation Analysis

```python
# Convert Gross to numeric if required
df["Gross"] = (
    df["Gross"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

df["Gross"] = pd.to_numeric(df["Gross"], errors="coerce")

# Select numerical columns
corr = df.select_dtypes(include="number").corr()

display(corr)

# Correlation with IMDb Rating
display(corr["IMDB_Rating"].sort_values(ascending=False))
```

# IMDb Rating vs Gross Revenue

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Gross"],
    df["IMDB_Rating"],
    alpha=0.5
)

plt.title("IMDb Rating vs Gross Revenue")
plt.xlabel("Gross Revenue")
plt.ylabel("IMDb Rating")

plt.show()
```
