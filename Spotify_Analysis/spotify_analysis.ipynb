# Spotify Analysis   

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Show all columns
pd.set_option("display.max_columns", None)

# Optional: Better looking tables in notebook
from IPython.display import display
```

## Load Dataset

```python
# Load dataset
df = pd.read_csv("C:\\Users\\91998\\Documents\\DataBooks\\Spotify_Analysis\\data\\spotify-tracks-dataset-detailed.csv")

print("Dataset loaded successfully!")
```

```python
print("Shape:", df.shape)

display(df.head())

display(df.tail())

df.info()
```

## Dataset Overview  

```python
print(df.isnull().sum())

print("Duplicates:", df.duplicated().sum())

display(df.describe())
```

## Data Cleaning  

```python
df = df.drop_duplicates()

if "Unnamed: 0" in df.columns:
    df.drop(columns=["Unnamed: 0"], inplace=True)

df.reset_index(drop=True, inplace=True)
```

```python
print(df.shape)
```

## Exploratory Data Analysis

```python
# Basic statistics
print(df["popularity"].describe())

# Most and least popular songs
print("\nTop 10 Most Popular Songs")
display(
    df[["track_name", "artists", "popularity"]]
    .sort_values(by="popularity", ascending=False)
    .head(10)
)

print("\nBottom 10 Least Popular Songs")
display(
    df[["track_name", "artists", "popularity"]]
    .sort_values(by="popularity")
    .head(10)
)
```

```python
plt.figure(figsize=(10,5))

plt.hist(df["popularity"], bins=20)

plt.title("Distribution of Song Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Songs")

plt.show()
```

```python
audio_features = [
    "danceability",
    "energy",
    "acousticness",
    "instrumentalness",
    "liveness",
    "speechiness",
    "valence",
    "tempo"
]

display(df[audio_features].describe())
```

```python
for feature in audio_features:
    plt.figure(figsize=(8,4))
    plt.hist(df[feature], bins=30)
    plt.title(f"{feature.capitalize()} Distribution")
    plt.xlabel(feature)
    plt.ylabel("Number of Songs")
    plt.show()
```

## Correlation Analysis

```python
numerical_df = df.select_dtypes(include=["number"])
```

```python
correlation = numerical_df.corr()

display(correlation)
```

```python
plt.figure(figsize=(12,8))

plt.imshow(correlation, cmap="coolwarm", aspect="auto")
plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=90)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
pop_corr = (
    correlation["popularity"]
    .sort_values(ascending=False)
)

display(pop_corr)
```

```python
features = [
    "danceability",
    "energy",
    "acousticness",
    "valence",
    "tempo",
    "speechiness",
    "instrumentalness",
    "liveness"
]

for feature in features:
    plt.figure(figsize=(7,4))

    plt.scatter(
        df[feature],
        df["popularity"],
        alpha=0.3,
        s=10
    )

    plt.title(f"Popularity vs {feature.capitalize()}")
    plt.xlabel(feature)
    plt.ylabel("Popularity")

    plt.show()
```

## Genre Analysis 

```python
top_genres = (
    df.groupby("track_genre")["popularity"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

display(top_genres)
```

```python
plt.figure(figsize=(10,5))

top_genres.plot(kind="bar")

plt.title("Top 10 Genres by Average Popularity")
plt.ylabel("Average Popularity")

plt.xticks(rotation=45)

plt.show()
```

```python
top_artists = (
    df.groupby("artists")["popularity"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

display(top_artists)
```

```python
plt.figure(figsize=(12,5))

top_artists.plot(kind="bar")

plt.title("Top Artists by Average Popularity")

plt.xticks(rotation=70)

plt.show()
```
